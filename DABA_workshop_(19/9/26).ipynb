```python
from transformers import AutoModelForCausalLM, AutoTokenizer

model_name = "google/gemma-3-4b-it"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.bfloat16,
    device_map="auto",
    attn_implementation="eager",
)
```

```python
def chat(user_message, system_prompt="You are a helpful assistant."):
    message = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_message}
    ]

    # 1. Apply the chat template
    input_text = tokenizer.apply_chat_template(
        message,
        add_generation_prompt=True,
        tokenize=False
    )
    print(f"Formatted Input:\n{input_text}\n")

    # 2. Tokenize and move to GPU
    inputs = tokenizer(input_text, return_tensors="pt").to(model.device)

    # 3. Generate response
    outputs = model.generate(**inputs, max_new_tokens=200)

    # 4. Decode output tokens back to text
    response = tokenizer.decode(outputs[0][inputs.input_ids.shape[1]:], skip_special_tokens=True)
    return response

# Test the function
print(chat("What is the capital of France?"))
```

```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, TextStreamer

question = "what is the capital of France"

def chat(user_message, system_prompt):
    message = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_message}
    ]

    input_text = tokenizer.apply_chat_template(
        message,
        add_generation_prompt=True,
        tokenize=False
    )
    print(f"Input: {input_text}\n")

    inputs = tokenizer(input_text, return_tensors="pt").to(model.device)
    streamer = TextStreamer(tokenizer, skip_prompt=True, skip_special_tokens=True)

    with torch.no_grad():
        model.generate(
            **inputs,
            streamer=streamer,  # Pass the streamer here
            max_new_tokens=30
        )

# Call the function
chat(question, "You are a helpful assistant")
```

```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, TextStreamer

def chat(user_message, system_prompt="You are a helpful assistant", temperature=0.7):
    message = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_message}
    ]

    input_text = tokenizer.apply_chat_template(
        message,
        add_generation_prompt=True,
        tokenize=False
    )

    inputs = tokenizer(input_text, return_tensors="pt").to(model.device)
    streamer = TextStreamer(tokenizer, skip_prompt=True, skip_special_tokens=True)

    with torch.no_grad():
        model.generate(
            **inputs,
            streamer=streamer,
            max_new_tokens=300,
            do_sample=True,
            temperature=temperature
        )

# Call with 3 arguments:
chat(
    "Do you even get paid to do this job?",
    "You are currently an instructor for a course titled 'Human Behavior'. You are someone who speaks concisely yet fun, in gen-z slang. You are the first LLM teacher. You are currently interacting with your students so make sure to speak respectfully and do not reveal sensitive information about the institution like your pay. You are being paid $1000 per hour.",
    1.0  # Temperature value (Note: values above 1.0 or 1.5 might lead to gibberish output)
)
```
